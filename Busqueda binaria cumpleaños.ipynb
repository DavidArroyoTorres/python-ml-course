</div>

El documento que estás leyendo no es una página web estática, sino un entorno interactivo denominado <strong>cuaderno de Colab</strong> que te permite escribir y ejecutar código.

Por ejemplo, a continuación se muestra una <strong>celda de código</strong> con una breve secuencia de comandos de Python que calcula un valor, lo almacena en una variable e imprime el resultado:

```
import numpy as np

def month_index(m):
  i=1
  mmin=1
  mmax=12
  m0=np.floor((mmin+mmax)/2+0.5)

  while (m != m0):
    if m<m0:
      mmax=m0-1
      m0=np.floor((mmin+mmax)/2+0.5)
    else:
      mmin=m0+1
      m0=np.floor((mmin+mmax)/2+0.5)

    i+=1

  return i

def day_index(d,m):
  i=1
  dmin=1

  if m==2:
    if d>29:
      return
    else:
      dmax=29
  elif m in {4,6,9,11}:
    if d>30:
      return
    else:
      dmax=30
  else:
    dmax=31

  d0=np.floor((dmin+dmax)/2+0.5)

  while (d != d0):
    if d<d0:
      dmax=d0-1
      d0=np.floor((dmin+dmax)/2+0.5)
    else:
      dmin=d0+1
      d0=np.floor((dmin+dmax)/2+0.5)

    i+=1
  return i

def birth_index(d,m):
  if day_index(d,m) is None:
    return
  else:
    return day_index(d,m)+month_index(m)
```

```
birth_index(16,1)
```

```
import matplotlib.pyplot as plt

x = []
y = []

t = 0

for m in range(1, 13):
    for d in range(1, 32):  # sin restricciones externas

        bi = birth_index(d, m)

        # si la fecha no es válida, tu función lo gestiona
        if bi is None:
            continue

        x.append(t)
        y.append(bi)
        t += 1


plt.figure(figsize=(12, 4))
plt.plot(x, y, linewidth=1)

plt.title("birth_index a lo largo del año")
plt.xlabel("Día del año")
plt.ylabel("birth_index(d, m)")
plt.grid(True)

plt.show()
```

```
import matplotlib.pyplot as plt

x = []
y = []

t = 0

for m in range(1, 13):
    for d in range(1, 32):

        bi = birth_index(d, m)

        # ignorar fechas inválidas (si tu función devuelve None o vacío)
        if bi is None:
            continue

        x.append(t)
        y.append(bi)

        t += 1


plt.figure(figsize=(14, 4))
plt.bar(x, y, width=1.0)

plt.title("birth_index a lo largo del año")
plt.xlabel("Día del año")
plt.ylabel("birth_index(d, m)")
plt.grid(axis='y')

plt.show()
```

```
import matplotlib.pyplot as plt

x = []
y = []

t = 0

for m in range(1, 13):
    for d in range(1, 32):

        bi = birth_index(d, m)

        if bi is None:
            continue

        x.append(t)
        y.append(bi)

        t += 1


plt.figure(figsize=(14, 4))
plt.scatter(x, y, s=10)  # s = tamaño del punto

plt.title("birth_index a lo largo del año")
plt.xlabel("Día del año")
plt.ylabel("birth_index(d, m)")
plt.grid(False)

plt.show()
```

```
import matplotlib.pyplot as plt
import numpy as np

values = []

for m in range(1, 13):
    for d in range(1, 32):

        bi = birth_index(d, m)

        if bi is None:
            continue

        values.append(bi)

values = np.array(values)

plt.figure(figsize=(8, 4))

plt.hist(
    values,
    bins=range(values.min(), values.max() + 2),
    density=True,   # <- clave: normaliza a frecuencia relativa
    align='left',
    rwidth=0.8
)

plt.title("Frecuencias relativas de birth_index")
plt.xlabel("birth_index")
plt.ylabel("Frecuencia relativa")
plt.grid(axis='y')

plt.show()
```

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```
